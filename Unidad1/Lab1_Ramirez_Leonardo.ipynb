In this lab you are going to explore and create a program that defines a class called Board

```python
class Board:
#Define the attributes of the class
  def __init__(self, color, material, size):
    self.color = color
    self.material = material
    self.size = size

#Define the methods of the class
  def write(self):
    print("Someone is writing in the board")
  def erase(self):
    print("The board has been erased")
  def describe(self):
    print(f"This board is color {self.color}, is made of {self.material}, and it has {self.size} squared meters of size")
#Self: se utiliza para decirle a python a que objeto pertencen los atributos

#Create an instance using the class "board"
board1 = Board("white", "tempered glass", "2m2")
board2 = Board("White", "Steel", "1.5m2")

#We access to the object (Instance) "table 1" to call its data
#Instance "table1"
print(board1.material)
print(board1.color)
board1.write()
board1.describe()

#Instance "table2"
print(board2.material)
```

# Lab 1: Bank account Class

Create a python program that models a bank account using Object-Oriented Programming.

```python
class BankAccount:

  def __init__(self, holder, __balance):
    self.holder = holder
    self.__balance = __balance

  def deposit(self,deposit):
    if deposit<=0:
      print("negative or zero amounts are not allowed")
    else:
      self.__balance+=deposit
    return(self.__balance)

  def withdrawal(self,withdrawal):
    if withdrawal>self.__balance:
      print("You can withdrawal more money than the account's balance")
    else:
      self.__balance-=withdrawal
    return (self.__balance)

  def check_balance(self,):
    print(f"The balance is of: ${self.__balance}")

account1 = BankAccount("Raul Perez", 5000)
account2 = BankAccount("Joel Lopez", 3000)

print(account1.holder)
account1.check_balance()
account1.deposit(300)
account1.check_balance()
account1.withdrawal(4000)
account1.check_balance()

print(account2.holder)
account2.check_balance()
account2.deposit(4300)
account2.check_balance()
account2.withdrawal(2600)
account2.check_balance()
```
